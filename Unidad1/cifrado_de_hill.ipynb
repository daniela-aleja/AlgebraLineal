# Evaluación de Álgebra Lineal en Computación: Cifrado de Hill
Este notebook contiene la implementación completa del **Cifrado de Hill** utilizando matrices de $2 \times 2$ y aritmética modular en $\mathbb{Z}_{26}$.

### Instrucciones:
1. Puedes modificar la matriz clave `K` en la primera celda de código.
2. Modifica la variable `mensaje_entrada` con el texto que desees cifrar.
3. Ejecuta las celdas para ver el proceso de cifrado y descifrado automatizado.

```python
# =====================================================================
# EVALUACIÓN DE ÁLGEBRA LINEAL - CARRERA DE COMPUTACIÓN
# TEMA: CIFRADO Y DESCIFRADO DE HILL CON MATRICES Y ARITMÉTICA MODULAR
# =====================================================================

import numpy as np

# ---------------------------------------------------------------------
# 1. DEFINICIÓN DE LA MATRIZ CLAVE (Modificable)
#    Puedes cambiar los valores de la matriz K (debe ser de 2x2 e invertible mod 26)
# ---------------------------------------------------------------------
K = np.array([
    [3, 5],
    [1, 2]
])

# Texto que se desea cifrar
mensaje_entrada = "HOLA"

# ---------------------------------------------------------------------
# 2. FUNCIONES AUXILIARES DE ÁLGEBRA LINEAL Y ARITMÉTICA MODULAR
# ---------------------------------------------------------------------
def inverso_modular(a, m=26):
    """Calcula el inverso multiplicativo de un entero 'a' modulo 'm'."""
    a = a % m
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    raise ValueError(f"El determinante {a} no tiene inverso en modulo {m}.")

def calcular_matriz_inversa_mod26(matriz):
    """Calcula la matriz inversa en modulo 26 usando la matriz adjunta."""
    # Determinante de la matriz 2x2: (a*d - b*c)
    det = int(np.round(np.linalg.det(matriz)))
    det_mod = det % 26
    
    # Inverso del determinante mod 26
    inv_det = inverso_modular(det_mod, 26)
    
    # Matriz Adjunta para matriz 2x2: [[d, -b], [-c, a]]
    adj = np.array([
        [matriz[1, 1], -matriz[0, 1]],
        [-matriz[1, 0],  matriz[0, 0]]
    ])
    
    # Inversa = inv_det * Adj(K) mod 26
    matriz_inversa = (inv_det * adj) % 26
    return matriz_inversa.astype(int)

# ---------------------------------------------------------------------
# 3. FUNCIONES DE CIFRADO Y DESCIFRADO
# ---------------------------------------------------------------------
def cifrar_hill(texto, clave):
    texto = texto.upper().replace(" ", "")
    if len(texto) % 2 != 0:
        texto += "X"  # Relleno si la cantidad de letras es impar
    
    texto_cifrado = ""
    for i in range(0, len(texto), 2):
        # Convertir par de letras a vector columna (2x1)
        v = np.array([[ord(texto[i]) - 65], [ord(texto[i+1]) - 65]])
        # Multiplicación matricial: C = K * P mod 26
        res = np.dot(clave, v) % 26
        texto_cifrado += chr(res[0, 0] + 65) + chr(res[1, 0] + 65)
        
    return texto_cifrado

def descifrar_hill(texto_cifrado, clave):
    # Obtener matriz inversa de la clave mod 26
    clave_inv = calcular_matriz_inversa_mod26(clave)
    texto_descifrado = ""
    
    for i in range(0, len(texto_cifrado), 2):
        # Convertir par cifrado a vector columna (2x1)
        v = np.array([[ord(texto_cifrado[i]) - 65], [ord(texto_cifrado[i+1]) - 65]])
        # Multiplicación matricial: P = K^(-1) * C mod 26
        res = np.dot(clave_inv, v) % 26
        texto_descifrado += chr(res[0, 0] + 65) + chr(res[1, 0] + 65)
        
    return texto_descifrado

# ---------------------------------------------------------------------
# 4. EJECUCIÓN Y SALIDA DE RESULTADOS
# ---------------------------------------------------------------------
K_inversa = calcular_matriz_inversa_mod26(K)
mensaje_cifrado = cifrar_hill(mensaje_entrada, K)
mensaje_descifrado = descifrar_hill(mensaje_cifrado, K)

print("=" * 55)
print("       DEMOSTRACIÓN DE CIFRADO DE HILL (ÁLGEBRA LINEAL)")
print("=" * 55)
print("Matriz Clave K (2x2):")
print(K)
print("\nMatriz Inversa K^(-1) mod 26:")
print(K_inversa)
print("-" * 55)
print(f"Texto Original:   {mensaje_entrada}")
print(f"Texto Cifrado:    {mensaje_cifrado}")
print(f"Texto Descifrado: {mensaje_descifrado}")
print("=" * 55)
```
